```python
import json
import pandas as pd
import os
import numpy as np
os.chdir('..')
```

```python
from copy import deepcopy
```

```python
lang = 'indo'
dataset_folder = 'corrected_splitopinion_typocorrected'
splits = ['train', 'test']
```

```python
dataset_per_split = {}
for split in splits:
	with open(f"hotel_dataset/{lang}/{dataset_folder}/hotel_aste_{split}_augmented{'_noreasoning' if split == 'train' else ''}.json", 'r') as f:
		dataset_per_split[split] = json.load(f)
```

```python
new_dataset_per_split = {}
# Only take the aos element order
for split in splits:
	new_dataset_per_split[split] = []
	for instance in dataset_per_split[split]:
		if instance['element_order'] == 'aos':
			instance['input'] = f"{instance['input']} =>" # Add arrow for input target delimiter (remove if not needed)
			new_dataset_per_split[split].append(deepcopy(instance))
```

```python
new_dataset_per_split['train'][370]
```

```python
# Save in json file
for split in splits:
	save_path = f"hotel_dataset/{lang}/{dataset_folder}_aos/hotel_aste_{split}_augmented{'_noreasoning' if split == 'train' else ''}.json"
	os.makedirs(os.path.dirname(save_path), exist_ok=True)
	with open(save_path, 'w') as f:
		json.dump(new_dataset_per_split[split], f, indent=4)
```
