```python
from copy import deepcopy
```

```python
with open('hotel_dataset/indo/clean_traintruncated/hotel_aste_train_augmented_noreasoning.json', 'r') as f:
	train_data_truth = json.load(f)
```

```python
with open('hotel_dataset/indo/corrected_splitopinion_typocorrected_aos/hotel_aste_train_augmented_noreasoning.json', 'r') as f:
	train_data = json.load(f)
```

```python
len(train_data_truth), len(train_data)
```

```python
# Take 5 step on the train_data_truth and store them to a new list
aos_train_data_truth = []
for instance in train_data_truth:
	if instance['element_order'] == 'aos':
		aos_train_data_truth.append(deepcopy(instance))
```

```python
invalid_indexes = []
for i in range(len(aos_train_data_truth)):
	if aos_train_data_truth[i]['input'] != train_data[i]['input']:
		invalid_indexes.append(i)
		print(f"--------- Invalid input: {i} ---------")
		for key, value in aos_train_data_truth[i].items():
			print(f"Truth {key}: {value}")
		for key, value in train_data[i].items():
			print(f"Data {key}: {value}")
		print("\n")
```

```python
for entry in train_data:
    if '[A] [O] [S]' not in entry['input']:
        print("Missing '[A] [O] [S]' in input:", entry)
```
